```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

from sklearn.preprocessing import StandardScaler, MinMaxScaler
```

```python
dados = pd.read_csv('data.csv', sep=",")
dados.head()
```

```python
x = dados[("sqft_living", "bathrooms")].values
y = dados["price"].values
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
x = dados[["sqft_living", "bathrooms"]].values
y = dados["price"].values
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, random_state=7)
```

```python
len(x_train)
```

```python
len(x_test)
```

```python
x_train
```

```python
scaler = MinMaxScaler()
scaler.fit(x_train)
```

```python
x_train_scaled = scaler.transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
x_train_scaled
```

```python
model = LinearRegression()
model.fit(x_train_scaled, y_train)
```

```python
y_pred = model.predict(x_test_scaled)
```

```python
MAE = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("MAE(min absolut error): ", MAE)
print("R2: ", r2)
```

```python
model_normal = LinearRegression()
model_normal.fit(x_train, y_train)
```

```python
y_pred_normal = model_normal.predict
```

```python
MAE = mean_absolute_error(y_test, y_pred_normal)
r2 = r2_score(y_test, y_pred_normal)

print("MAE(min absolut error): ", MAE)
print("R2: ", r2)
```
