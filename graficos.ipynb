```python
import pandas as pd
```

```python
df = pd.read_csv("canadian_immegration_data.csv")
```

```python
df
```

```python
df.info()
```

```python
df.set_index("Country", inplace=True)
```

```python
anos = list(map(str, range(1980, 2014)))
```

```python
anos
```

```python
brasil = df.loc["Brazil", anos]
```

```python
brasil
```

```python
brasil_dict = {"ano" : brasil.index.tolist(), "imigrantes" : brasil.values.tolist()}
dados_brasil = pd.DataFrame(brasil_dict)
```

```python
dados_brasil
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,4))
plt.plot(dados_brasil["ano"], dados_brasil["imigrantes"])
plt.title("Imigraçao do Brasil para Canadá")
plt.xlabel("Ano")
plt.ylabel("Quantidade")
plt.xticks(["1980", "1985", "1990", "1995", "2000", "2005", "2010"])
plt.show()
```

```python
fig, ax = plt.subplots(figsize = (8,4))
ax.plot(dados_brasil["ano"], dados_brasil["imigrantes"])
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
plt.show()
```

```python
fig, axs = plt.subplots(1, 2, figsize=(15,5))

axs[0].plot(dados_brasil["ano"], dados_brasil["imigrantes"])
axs[0].set_title("imigração do brasil para o canada\n1980 a 2013")
axs[0].set_xlabel("Ano")
axs[0].set_ylabel("Quantidade")
axs[0].xaxis.set_major_locator(plt.MultipleLocator(5))
axs[0].grid()

axs[1].boxplot(dados_brasil["imigrantes"])
axs[1].set_title("Boxplot da quantidade de imigrantes do brasil para o canada\n1980 a 2013")
axs[1].set_xlabel("Brasil")
axs[1].set_ylabel("Quantidade")
axs[1].grid()

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8,4))
ax = sns.barplot(data=top_10, x='Total', y=top_10.index, orient='h')

ax.set_title('paises com maior imigração\n1980 a 2013', loc='left', fontsize=16)
ax.set_xlabel('numero de imigrantes', fontsize=14)
ax.set_ylabel('')

plt.show()
```

```python
def gerar_paleta(palette):
  fig, ax = plt.subplots(figsize=(8,4))
  ax = sns.barplot(data=top_10, x='Total', y=top_10.index, orient='h', palette=palette)

  ax.set_title('paises com maior imigração\n1980 a 2013', loc='left', fontsize=18)
  ax.set_xlabel('QTDE', fontsize=14)
  ax.set_ylabel('')

  plt.show()
```
