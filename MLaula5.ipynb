```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("airlines.csv", sep=',')
df.head()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
#valores nulos
df = df.dropna()
```

```python
duplicated_cols = []
for col in df.columns:
    if df[col].duplicated().any():
        duplicated_cols.append(col)
print(duplicated_cols)
```

```python
df.describe()
```

```python
fig, ax = plt.subplots(figsize=(8,6))
sns.violinplot(x='Lenght', data=df, ax=ax, color='lightgray')
sns.boxplot(x='Lenght', data=df, ax=ax, whis=1.5, color='darblue')
ax.set_title('boxplot e violinplot')
plt.show()
```
