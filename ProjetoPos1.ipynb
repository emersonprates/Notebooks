```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
import seaborn as sns
```

```python
df = pd.read_csv("Exportacao.csv", sep=';')
df2 = pd.read_csv("Importacao.csv", sep=';')
```

```python
df.head(23)
```

```python
df2.head(13)
```

```python
df.set_index("País", inplace=True)
```

```python
anos = list(map(str, range(2009, 2024)))
```

```python
anos
```

```python
brasil = df.loc["Brasil", anos]
brasil
```

```python
brasil_dict = {"anos" : brasil.index.tolist(), "valor" : brasil.values.tolist()}
dados_brasil = pd.DataFrame(brasil_dict)
```

```python
dados_brasil
```

```python
plt.figure(figsize=(12,4))
plt.plot(dados_brasil["anos"], dados_brasil["valor"])
plt.xlabel("Ano")
plt.ylabel("Quantidade")
plt.xticks(["2009", "2011", "2013", "2015", "2017", "2019", "2021", "2023"], fontsize=10)
plt.show()
```

```python
#apply(pd.to_numeric, errors='coerce') converte todos valores para numerico e coerce força essa conversão para numerico e mostra o que não é como null
df_anos = df[anos].apply(pd.to_numeric, errors='coerce')

total_por_ano = df_anos.sum(axis=0)

print(total_por_ano)
```

```python
#total exportados nos anos
fig, ax = plt.subplots(figsize=(12,8))
sns.barplot(x=total_por_ano.index, y=total_por_ano.values)
```

```python
fig, ax = plt.subplots(figsize=(12,8))
ax = sns.barplot(x=total_por_ano.index, y=total_por_ano.values)
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
#plt.xticks(rotation=90)

fig.savefig("qtdeano.jpg")

plt.show()
```

```python
total_top = total_por_ano.sort_values(ascending=False).head(5)
total_top
```

```python
#anos que mais exportaram (qtde)
fig, ax = plt.subplots(figsize=(10,6))
sns.barplot(x=total_top.index, y=total_top.values)

fig.savefig("topanos.jpg")
```

```python
total_por_pais = df_anos.sum(axis=1)
total_por_pais.head(23)
```

```python
total_por_pais_top = total_por_pais.sort_values(ascending=False).head(5)
total_por_pais_top
```

```python
#tabela apenas contendo Paraguai, Estados Unidos, Rússia, Japão e Angola

top_paises = ["Rússia", "Paraguai", "Estados Unidos", "China", "Haiti"]
df_top_paises = df.loc[top_paises, anos].apply(pd.to_numeric, errors='coerce')
df_top_paises
```

```python
#paises que mais receberam a exportaçao (dinamico)
fig = px.line(df_top_paises, x=df_top_paises.index, y=df_top_paises.columns,
              title='Top 5 países', markers=True)
fig.show()
```

```python
#trocar os países pelos anos e a legenda pelos paises do grafico acima
#paises que mais receberam a exportaçao (dinamico)
fig = px.line(df_top_paises.T, x=df_top_paises.T.index, y=df_top_paises.T.columns,
              title='Top 5 países', markers=True)
fig.update_layout(xaxis={'tickangle':-45})
fig.show()

fig.write_html("toppaisesint.html")
```

```python
#paises que mais receberam a exportaçao
cores = ['navy', 'blue', 'royalblue', 'cornflowerblue', 'lightsteelblue']
fig, ax = plt.subplots(figsize=(10,6))
sns.barplot(x=total_por_pais_top.index, y=total_por_pais_top.values, palette=cores)

fig.savefig("toppaises.jpg")
```
